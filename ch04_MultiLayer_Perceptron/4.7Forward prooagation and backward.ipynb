# 前向传播、反向传播和计算图

## 前向传播
按照输入层到输出层的顺序，计算和存储神经网络中的每一层的结果。假设输入样本是$x \in \mathbb{R}^{d}$，隐藏层中不包括偏置项，中间变量是：
$$
z = W^{(1)}x
$$

其中$W^{(1)} \in \mathbb{R}^{h \times d}$，$h$是隐藏层的神经元数量。中间变量$z \in \mathbb{R}^{h} $通过激活函数 $\Phi$后，得到长度为h的隐藏激活向量：
$$
h = \Phi(z)
$$

隐藏变量$h$是一个中间变量。假设输出层的参数只有权重$W^{(2)} \in \mathbb{R}^{q \times h}$，可以得到输出层变量，是一个长度为q的向量：
$$
o = W^{(2)}h
$$


假设损失函数是l，样本标签为y，计算单个数据样本的损失项：
$$
L = l(o, y)
$$


根据L2正则化的定义，给定超参数$\lambda$，正则化项为：
$$
s = \frac{\lambda}{2} (||W^{(1)}||^2_{F} + ||W^{(2)}||^2_{F})
$$


其中矩阵的Frobenius范数是把矩阵展平之后应用的L2范数。模型在给定数据样本上的正则化损失为：
$$
J = L + s
$$

## 反向传播

反向传播是为了计算神经网络参数的梯度 。在这个带有单隐藏层和 $L_2$ 正则化的多层感知机模型中，我们的最终目标是计算目标函数 $J$ 对隐藏层权重参数 $\mathbf{W}^{(1)}$ 和输出层权重参数 $\mathbf{W}^{(2)}$ 的梯度，即分别求出 $\frac{\partial J}{\partial \mathbf{W}^{(1)}}$ 和 $\frac{\partial J}{\partial \mathbf{W}^{(2)}}$ 。

各个参数如下：
输入向量 $\mathbf{x}$: $\mathbb{R}^{d \times 1}$ （长度为$d$的列向量）

隐藏层权重矩阵 $\mathbf{W}^{(1)}$: $\mathbb{R}^{h \times d}$ （$h$是隐藏层单元数）

中间变量 $\mathbf{z}$: $\mathbf{z} = \mathbf{W}^{(1)}\mathbf{x}$，维度为 $\mathbb{R}^{h \times 1}$

隐藏层激活值 $\mathbf{h}$: $\mathbf{h} = \phi(\mathbf{z})$，维度为 $\mathbb{R}^{h \times 1}$

输出层权重矩阵 $\mathbf{W}^{(2)}$: $\mathbb{R}^{q \times h}$ （$q$是输出类别的数量）

输出变量 $\mathbf{o}$: $\mathbf{o} = \mathbf{W}^{(2)}\mathbf{h}$，维度为 $\mathbb{R}^{q \times 1}$

损失项 $L$ 与 正则项 $s$: 两者都是标量，维度为 $\mathbb{R}^{1 \times 1}$

目标函数 $J$: $J = L + s$，同样是标量 $\mathbb{R}^{1 \times 1}$

在机器学习的矩阵求导中，标量对某个矩阵或向量求导，其梯度的维度与该矩阵或向量的维度完全一致，例如，$\frac{\partial J}{\partial \mathbf{W}^{(2)}}$ 的维度必须与 $\mathbf{W}^{(2)}$ 一样，是 $\mathbb{R}^{q \times h}$ 

### 计算靠近输出层的参数 $\mathbf{W}^{(2)}$ 的梯度

标量目标函数 $J$ 对输出向量 $\mathbf{o}$ 的梯度 $\frac{\partial J}{\partial \mathbf{o}}$ 维度为 $\mathbb{R}^{q \times 1}$ 。
我们需要求 $\frac{\partial J}{\partial \mathbf{W}^{(2)}}$，其维度必须是 $\mathbb{R}^{q \times h}$

已知前向传播公式为：$\mathbf{o} = \mathbf{W}^{(2)}\mathbf{h}$

根据链式法则，我们需要将 $\frac{\partial J}{\partial \mathbf{o}}$（维度 $q \times 1$）和 $\mathbf{h}$（维度 $h \times 1$）结合起来，产生一个 $q \times h$ 的矩阵。在线性代数中，要让一个列向量 $(q \times 1)$ 和另一个列向量 $(h \times 1)$ 相乘得到一个矩阵 $(q \times h)$，唯一的方法是进行外积（Outer Product），即将第二个向量转置为行向量：$$(q \times 1) \times (1 \times h) = (q \times h)$$

已知目标函数 $J = L + s$（其中 $L$ 为损失项，$s$ 为正则化项），第一步是计算目标函数 $J$ 相对于 $L$ 和 $s$ 的梯度 ：
  $$\frac{\partial J}{\partial L} = 1 \quad \text{且} \quad \frac{\partial J}{\partial s} = 1$$

根据链式法则，目标函数关于输出层变量 $\mathbf{o}$ 的梯度计算如下(该梯度的维度为 $\mathbb{R}^{q \times 1}$ 。) ：
  $$\frac{\partial J}{\partial \mathbf{o}} = \text{prod}\left(\frac{\partial J}{\partial L}, \frac{\partial L}{\partial \mathbf{o}}\right) = \frac{\partial L}{\partial \mathbf{o}}$$

计算正则化项对模型权重的梯度
正则化项 $s = \frac{\lambda}{2} \left( \|\mathbf{W}^{(1)}\|_F^2 + \|\mathbf{W}^{(2)}\|_F^2 \right)$ 对两个权重参数矩阵的梯度分别为 ：
  $$\frac{\partial s}{\partial \mathbf{W}^{(1)}} = \lambda \mathbf{W}^{(1)} \quad \text{且} \quad \frac{\partial s}{\partial \mathbf{W}^{(2)}} = \lambda \mathbf{W}^{(2)}$$
